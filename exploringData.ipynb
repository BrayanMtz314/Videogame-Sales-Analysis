```python
def dat_descriptivos(df):
    """obtiene  datos descriptivos por medio de agreagaciones"""
    print("--------------------- DATOS DESCRIPTIVOS -------------------------")
    resumen = df[["Global_Sales", "Other_Sales", "JP_Sales","NA_Sales", "EU_Sales"]].agg(["mean", "median", "sum", "var", "std", "kurt"])
    print(resumen)
    print()
```
